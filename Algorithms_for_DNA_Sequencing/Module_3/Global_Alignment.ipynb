```python
alphabet = ['A', 'C', 'G', 'T']
# Penalty for Transition = 2, Transversion = 4, Gap (INDEL)= 8, Match = 0

# Penalty Matrix Below as 5 by 5 array
score = [[0, 4, 2, 4, 8],     # For 'A' against A,C,G,T and Gap   
         [4, 0, 4, 2, 8],     # For 'C' against A,C,G,T and Gap
         [2, 4, 0, 4, 8],     # For 'G' against A,C,G,T and Gap
         [4, 2, 4, 0, 8],     # For 'T' against A,C,G,T and Gap
         [8, 8, 8, 8, 8]]
```

```python
def GlobalAlignment(x,y):
    D = []  #Matrix D
    for i in range(len(x)+1):   # +1 is for consideration of empty string
        D.append([0] * (len(y)+1))

    #Inititalize the first row and first Column with the penalty of skip (8 here)
    for i in  range(1, len(x)+1):            # Considering from 1 as top left cell in a matrix will always be zero
        D[i][0] = D[i-1][0] + score[alphabet.index(x[i-1])][-1]  # Fill the 1st column of matrix D by taking penalty from the cell of a row above (D[i-1][0]) and adding gap penalty from score matrix (alphabet.index(x[i-1]) returns the index of a current char (i-1 = index correction for x as we start from 1) to look for the row in the score matrix, -1 is for last value of a row that contains gap penalty)
    for i in range(1,len(y)+1):
        D[0][i] = D[0][i-1] + score[-1][alphabet.index(y[i-1])]  # Fill the 1st row of matrix D by taking penalty from the cell of a column left to it (D[0][i-1]) and adding gap penalty from score matrix (-1 is for last row that contains gap penalty, alphabet.index(y[i-1]) returns the index of a current char (i-1 = index correction for y as we start from 1) to look for the column in the score matrix)

    #Loop through each row for each column
    for i in range(1, len(x)+1):
        for j in range(1, len(y)+1):
            distHor = D[i][j-1] + score[-1][alphabet.index(y[j-1])] #D[i][j-1] (any value in row i and column j-1) + gap penalty for skipping char in y[j-1]   
            distVer = D[i-1][j] + score[alphabet.index(x[i-1])][-1] #D[i-1][j] (any value in row i-1 and column j) + gap penalty for skipping char in x[i-1]
            #for Diagonal
            if x[i-1] == y [j-1]:
                distDiag = D[i-1][j-1]
            else:
                distDiag = D[i-1][j-1] + score[alphabet.index(x[i-1])][alphabet.index(y[j-1])] #D[i-1][j-1] + penalty from score matrix for substituting x[i-1] with y[j-1]
            D[i][j] = min(distHor, distVer, distDiag)
    return D[-1][-1]      
```

```python
X = "TGCGTAGTGCGT"
Y = "TGCCTAGGTGCGT"
print(GlobalAlignment(X,Y))
```
