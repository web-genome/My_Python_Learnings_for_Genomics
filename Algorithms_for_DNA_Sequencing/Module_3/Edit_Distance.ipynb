```python
def editDistrecursive(x,y):
    #slow algorithm
    if len(x) == 0:
        return len(y)
    elif len(y) == 0:
        return len(x)
    else:
        distHor = editDistrecursive(x[:-1], y) + 1 #Hor = Horizontal
        distVer = editDistrecursive(x, y[:-1]) + 1 #Ver =Vertical
        if x[-1] == y[-1]:
            distDiag = editDistrecursive(x[:-1], y[:-1]) #Diag = Diagonal
        else:
            distDiag = editDistrecursive(x[:-1], y[:-1]) + 1
        return min(distHor, distVer, distDiag)
    
```

```python
def editDistance(x,y):
    D = []  # Matrix D
    for i in range(len(x)+1):   # +1 is for consideration of empty string
        D.append([0] * (len(y)+1)) # Create 2D array with zeros

    # Inititalize the first row and first Column from 0 to len of strings(i.e., increment with the penalty of 1 as we move right (Y) or down (X))
    for i in  range(len(x)+1):
        D[i][0] = i
    for i in range(len(y)+1):
        D[0][i] = i

    # Loop through each remaining row for each corresponding remaining column and fill in the rest of the matrix
    for i in range(1, len(x)+1):
        for j in range(1, len(y)+1):
            distHor = D[i][j-1] + 1     # Here +1 designates penalty for skipped char in column y (shows one char remained to be considered)
            distVer = D[i-1][j] + 1
            #for Diagonal
            if x[i-1] == y [j-1]:
                distDiag = D[i-1][j-1]
            else:
                distDiag = D[i-1][j-1] + 1
            D[i][j] = min(distHor, distVer, distDiag)

    # Edit distance is the value in the bottom right corner of the matrix
    return D[-1][-1]      
```

```python
%%time  #returs runtime to execute the code block
x = "Vv Loves Pa"
y = "vv lovespa"
print(editDistrecursive(x,y))

# Takes longer time to execute as it is a recursive function without memoization, which leads to repeated calculations for the same inputs.
```

```python
%%time
x = "Vv Loves Pa"
y = "vv lovespa"
print(editDistance(x,y))

# The dynamic programming approach is significantly faster than the recursive approach, especially for longer strings, as it avoids redundant calculations by storing previously computed results in a matrix.
```
