```python
def overlap(a,b, min_length=3):
    """Finds overlaps between string a and b where 
    suffix of 'a' overlaps prefix of 'b' with minimum length of 3
    and returns the length of that overlap"""
    
    start = 0  # Initialize the index of 'a' with zero.

    while True:
        start = a.find(b[:min_length], start)       # Update the index of 'a' as it move forward to find prefix of 'b' in 'a'
        if start == -1:             # Find function returns -1, if given argument (here prefix of 'b') not found.
            return 0
        if b.startswith(a[start:]): # Verifies if the prefix of 'b' is equal to suffix of 'a' over the rest of the lenght from starting at the position 'start'.
            return len(a)-start     # Returns length of the overlap
        start += 1                  # Increament an index to start with
```

```python
from itertools import permutations
list(permutations(['a','b'],2)) # Just to show how it works
```

```python
def naive_overlap_map(reads,k):
    olaps = {}      # Overlap map as dictionary
    for a,b in permutations(reads,2):       # Assign reads in purmutations to 'a' and 'b'. eg., reads = X and Y. First permutation = (X=a, Y=b) and Second Permutation = (Y=a, X=b)
        olen = overlap(a,b, min_length=k)   # Use overlap function to find length of the overlap
        if olen > 0:
            olaps[(a,b)] = olen             # If overlap, store reads in tuple as a key of a dict and len as values  
    return olaps
```

```python
reads = ['ACGGATGATC', "GATCAAGT", "TTCACGGA"]
print(naive_overlap_map(reads, 3))
```
