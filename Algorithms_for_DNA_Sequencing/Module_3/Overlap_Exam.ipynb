```python
def Readfastq(filename):
    sequences = []
    
    with open(filename) as f:
        while True:
            f.readline()                #ignore first line of a read i.e., Name
            seq = f.readline().rstrip() # removes white space while reading only add sequence
            f.readline()                #ignore third line of a read i.e., placeholder strand (+ or -)
            f.readline()                #ignore fourth line of a read i.e., qualities
            if len(seq) == 0:
                break
            sequences.append(seq)
            
    return sequences

Reads = Readfastq(r"F:\Python\Algorithms\Module 3\ERR266411_1.for_asm.fastq")

```

```python
def overlap(a,b, min_length):
    """Finds overlaps between string a and b where 
    suffix of 'a' overlaps prefix of 'b' with minimum length of k
    and returns the length of that overlap"""
    
    start = 0  # Initialize the index of 'a' with zero.

    while True:
        start = a.find(b[:min_length], start)       # Update the index of 'a' as it move forward to find prefix of 'b' in 'a'
        if start == -1:             # Find function returns -1, if given argument (here prefix of 'b') not found.
            return 0
        if b.startswith(a[start:]): # Verifies if the prefix of 'b' is equal to suffix of 'a' over the rest of the lenght from starting at the position 'start'.
            return len(a)-start     # Returns length of the overlap
        start += 1                  # Increament an index to start with
```

```python
def kmer_and_read(reads, k):
    """Returns a dictionary with kmers as keys and set of reads containing that kmer as values"""
    kmers_with_reads = {}
    for seq in reads:
        for i in range(len(seq) - k + 1):
            kmer = seq[i:i+k]
            if kmer not in kmers_with_reads:
                kmers_with_reads[kmer] = set()
            kmers_with_reads[kmer].add(seq)

    return kmers_with_reads


```

```python
def overlap_all_pairs(reads, k):
    """Finds all overlaps between all pairs of reads with minimum overlap length k
    and returns a dictionary with keys as tuple of reads and values as length of overlap"""

    kmers_with_reads = kmer_and_read(reads, k) #Get kmers with reads dictionary
    olaps = {}
    for a in reads:           # Iterate through all reads   
        suffix = a[-k:]       # Get suffix of length k from read 'a'
        if suffix in kmers_with_reads:          # check if suffix of 'a' is present in the kmers_with_reads dictionary
            for b in kmers_with_reads[suffix]:  # Iterate through all reads containing that kmer
                if a != b:                      # Avoid comparing same reads
                    olen = overlap(a,b, min_length=k)   # Use overlap function to find length of the overlap
                    if olen > 0:
                        olaps[(a,b)] = olen # Store the overlap length in the dictionary with key as tuple of reads (a,b)
                    
    return olaps
```

```python
Edges = overlap_all_pairs(Reads,30)
Nodes = set()       # To store unique nodes having the outgoing edges (suffix involved in overlapping)
for a,b in Edges.keys():
    Nodes.add(a)
print(len(Nodes))
print(len(Edges))
```

```python
from itertools import permutations

def naive_overlap_map(reads,k):
    olaps = {}      # Overlap map as dictionary
    for a,b in permutations(reads,2):       # Assign reads in purmutations to 'a' and 'b'. eg., reads = X and Y. First permutation = (X=a, Y=b) and Second Permutation = (Y=a, X=b)
        olen = overlap(a,b, min_length=k)   # Use overlap function to find length of the overlap
        if olen > 0:
            olaps[(a,b)] = olen             # If overlap, store reads in tuple as a key of a dict and len as values  
    return olaps
print(len(naive_overlap_map(Reads, 30)))
```
