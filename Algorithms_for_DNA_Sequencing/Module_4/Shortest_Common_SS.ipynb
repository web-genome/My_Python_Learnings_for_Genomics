```python
def overlap(a,b, min_length):
    """Finds overlaps between string a and b where 
    suffix of 'a' overlaps prefix of 'b' with minimum length of k
    and returns the length of that overlap"""
    
    start = 0  # Initialize the index of 'a' with zero.

    while True:
        start = a.find(b[:min_length], start)       # Update the index of 'a' as it move forward to find prefix of 'b' in 'a'
        if start == -1:             # Find function returns -1, if given argument (here prefix of 'b') not found.
            return 0
        if b.startswith(a[start:]): # Verifies if the prefix of 'b' is equal to suffix of 'a' over the rest of the lenght from starting at the position 'start'.
            return len(a)-start     # Returns length of the overlap
        start += 1                  # Increament an index to start with
```

```python
import itertools
def scs(ss):
    """Given the lists of strings, returns shortest superstring
    containing all those strings as a substring from all the possible orders (permutations)"""

    shortest_sup = None                     # Set the shortest_sup to null
    total_SCS_observed = 0                  # Keep track of nunber of SCS tried
    tried_sup_list = []
    total_SCS_tried = [] 

    for ssperm in itertools.permutations(ss):       # Iterate through every order in a permutations of a given set of strings as list (ss)
        sup = ssperm[0]                             # Assign the initial value to a 'sup' variable as a first string of a current order
        

        for i in range(len(ss)-1):                  # Iterate though remaining strings other than first string
            olen = overlap(ssperm[i], ssperm[i + 1], min_length=1)      # Determine the overlap length between the suffix of a current string and prefix of a next string in a loop
            sup += ssperm[i+1][olen:]               # Append the non-overlapping character(s) of the next string to a current 'sup' variable updating the sup at each iteration
     
               
        if shortest_sup is None or len(sup) < len(shortest_sup):    # Check the length of the 'sup' variable of current iteration against the shortest_sup (None if 1st order in an permutation iteration) 
            shortest_sup = sup                      # Update the shortest superstring if len is shorter
        tried_sup_list.append(sup) 

    for item in tried_sup_list:                     # Track total SCS tried
        if len(item) == len(shortest_sup):
            total_SCS_tried.append(item) 

            
    print(f'Total number of SCS found are {total_SCS_observed}.')                   
    return shortest_sup, total_SCS_tried                             
```

```python
Result = scs(['GAT', 'TAG', 'TCG', 'TGC', 'AAT', 'ATA'])
print(Result)
```

```python
def pick_maximal_overlap(reads,k):
    """Returns a pair of reads with maximum overlap among
    all pairs along with thier length, k refers to miminum overlap
    to consider while comparing the length"""

    read_a, read_b = None, None # To store the pair of reads having maximal overlap
    best_olen = 0       # Variable to track best overlap initialized to zero

    for a,b in itertools.permutations(reads, 2):    # Take every possible pair of reads
        olen = overlap(a, b, min_length = k)        # Calculate the overlap length between the given pair of reads
        if olen > best_olen:                        # Compare the current Olap length with best Olap length
            read_a, read_b = a, b                   # If Olap is larger, assign current reads in a pair to the variables read_a and read_b
            best_olen = olen
    return read_a, read_b, best_olen                # If there are ties between pairs in terms of Olap length, then it just returns the first pair encountered since o_len must be > best_olen


```

```python
def greedy_scs(reads, k):
    """Returns the shortest common superstring by making decisions based on the 
    maximum overlap between reads (meaning shortest common string for a given pairs of reads),
    concatinating those reads without the repeat of an overlap and again checking 
    for the maximum overlap"""

    read_a, read_b, olen = pick_maximal_overlap(reads,k)    # Returns the pair of reads having maximum overlap among all reads
    
    while olen > 0:         # Loop through until there is no edges between the nodes (overlap)
        
        reads.remove(read_a)   # remove the pair having maximum Olap (read_a and read_b) from the list of reads
        reads.remove(read_b)

        reads.append(read_a + read_b[olen:])    # Append the resulting read by overlapping read_a and read_b to the list of read
        read_a, read_b, olen = pick_maximal_overlap(reads, k)   # Again check for the pair with maximum Olap with this new list of strings
    
    return ''.join(reads)   # Join the remaining reads (nodes) having no edges in between
```

```python
greedy_scs(['ABC', 'BCA', 'CAB'], 2)
```

```python
print(greedy_scs(['ABCD', 'CDBC', 'BCDA'], 1))  # Faster than SCS, but can return longer SCS than true SCS
print(len(greedy_scs(['ABCD', 'CDBC', 'BCDA'], 1)))
```
